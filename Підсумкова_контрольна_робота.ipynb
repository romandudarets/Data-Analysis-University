```python
import pandas as pd
import matplotlib.pyplot as plt

df1 = pd.read_csv('/content/Global_YouTube_Statistics[1].csv', encoding='latin1')

print(df1.head())

print(df1.shape)
print(df1.isnull().sum())
print(df1.duplicated().sum())

num_cols = df1.select_dtypes(include=['number']).columns
df1[num_cols] = df1[num_cols].astype(float)
df1[num_cols] = df1[num_cols].apply(lambda x: x.fillna(x.mean()))

print(df1.isnull().sum())

print(df1['Country'].nunique())

df1['video views'].plot(kind='hist', bins=20)
plt.show()

print(df1['video views'].max())
print(df1['video views'].min())
print(df1['video views'].mean())

print(df1.groupby('Country')['uploads'].sum().idxmax())

print(df1.loc[df1['uploads'].idxmax(), 'Title'])
print(df1.loc[df1['uploads'].idxmin(), 'Title'])
```

```python
df2 = pd.read_csv('shopping_trends.csv')

print(df2.shape)
print(df2.isnull().sum())
print(df2.dtypes)

df2.columns = df2.columns.str.lower().str.replace(' ', '_')

df2['review_rating'].plot(kind='hist', bins=10)
plt.show()

df2['gender'].value_counts().plot(kind='bar')
plt.show()

df2.groupby('gender')['purchase_amount_(usd)'].sum().plot(kind='bar')
plt.show()

df2.plot.scatter(x='review_rating', y='purchase_amount_(usd)')
plt.show()
```

```python
import altair as alt
from vega_datasets import data
import plotly.express as px

wheat = data.wheat()
bars = alt.Chart(wheat).mark_bar().encode(
    x='wheat:Q',
    y='year:O'
)
text = bars.mark_text(
    align='left',
    baseline='middle',
    dx=3
).encode(
    text='wheat:Q'
)
chart1 = bars + text
chart1.display()

iowa = data.iowa_electricity()
chart2 = alt.Chart(iowa).mark_bar(opacity=0.7).encode(
    x='year:O',
    y='net_generation:Q',
    color='source:N'
)
chart2.display()

iris = px.data.iris()
fig = px.scatter(iris, x='sepal_width', y='sepal_length')
fig.show()
```
