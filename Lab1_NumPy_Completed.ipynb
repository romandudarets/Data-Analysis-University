### Завдання 25
Обчисліть косинус кута та кут у градусах між векторами `a=[1, 2]` і `b=[2, 1]`.

```python
a25 = np.array([1.0, 2.0])
b25 = np.array([2.0, 1.0])

cos_theta = (a25 @ b25) / (np.linalg.norm(a25) * np.linalg.norm(b25))
cos_theta_clipped = np.clip(cos_theta, -1.0, 1.0)
angle_deg = np.degrees(np.arccos(cos_theta_clipped))

print("cos(theta):", cos_theta_clipped)
print("Кут у градусах:", angle_deg)
```

Для матриці `A=[[4, 7], [2, 6]]` знайдіть визначник, обернену матрицю та перевірте добуток `A @ inv(A)`.

```python
A28 = np.array([[4.0, 7.0], [2.0, 6.0]])
det_A = np.linalg.det(A28)
inv_A = np.linalg.inv(A28)
check_id = A28 @ inv_A

print("Визначник:", det_A)
print("Обернена матриця:\n", inv_A)
print("Перевірка A @ inv(A):\n", check_id)
```

### Завдання 37

```python
np.random.seed(42)
mat37 = np.random.randint(1, 21, size=(5, 5))

main_diag_sum = np.trace(mat37)
anti_diag_sum = np.trace(np.fliplr(mat37))
row_max = np.max(mat37, axis=1)
col_min = np.min(mat37, axis=0)

mean_all = np.mean(mat37)
mat37_filtered = np.where(mat37 < mean_all, 0, mat37)

print("Початкова матриця 5x5:\n", mat37)
print("1. Сума головної діагоналі:", main_diag_sum)
print("2. Сума побічної діагоналі:", anti_diag_sum)
print("3. Максимум по кожному рядку:", row_max)
print("4. Мінімум по кожному стовпцю:", col_min)
print(f"5. Матриця після заміни елементів < {mean_all:.2f} на 0:\n", mat37_filtered)
```

```python
A40 = np.array([
    [1.0,  2.0, 1.0],
    [2.0,  1.0, -1.0],
    [3.0, -1.0, 2.0]
])
b40 = np.array([9.0, 8.0, 3.0])

x40 = np.linalg.solve(A40, b40)
check40 = A40 @ x40

print("Розв'язок [x, y, z]:", x40)
print("Перевірка A @ x:", check40)
print("Вектор b:", b40)
print("Збігається з b:", np.allclose(check40, b40))
```
