### Завдання 5
Створіть два одновимірних масиви розміром 6, заповніть їх випадковими цілими числами від 1 до 20 та виконайте операції.

```python
# 5. Операції над векторами
a = np.random.randint(1, 21, size=6)
b = np.random.randint(1, 21, size=6)
print(f'a: {a}')
print(f'b: {b}')
print('Додавання:', a + b)
print('Віднімання:', a - b)
print('Множення:', a * b)
```

### Завдання 6
Створіть дві матриці розміром 3x2 та 2x4 і перемножте їх.

```python
# 6. Множення матриць
matrix_a = np.random.randint(1, 6, (3, 2))
matrix_b = np.random.randint(1, 6, (2, 4))
product = np.dot(matrix_a, matrix_b)
print('Результат множення:\n', product)
```

### Завдання 7

### Завдання 8

### Завдання 9
Дві матриці 3x3 (від -5 до 5), знайдіть добуток.

```python
# 9. Добуток матриць
m3 = np.random.randint(-5, 6, (3, 3))
m4 = np.random.randint(-5, 6, (3, 3))
print('Добуток:\n', np.dot(m3, m4))
```

### Завдання 10
Матриця 3x3 (від -1 до 1), вектор-стовпчик сум рядків.

```python
# 10. Сума рядків
last_matrix = np.random.uniform(-1, 1, (3, 3))
row_sums = last_matrix.sum(axis=1).reshape(-1, 1)
print('Вектор-стовпчик сум рядків:\n', row_sums)
```
